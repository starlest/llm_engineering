from agents.deals import Deal, Opportunity

    def __init__(self):

                      url="https://merch.example/l/101")
        self.memory = [Opportunity(deal=figure, estimate=90.0, discount=44.5)]

    def run(self):
        program = Deal(product_description="WrestleMania III event program, 1987", price=60.0,
                       url="https://merch.example/l/103")
        self.memory.append(Opportunity(deal=program, estimate=112.25, discount=52.25))
        return self.memory


class PretendMessenger:
